```python
import pandas as pd
import numpy as np
from collections import Counter
import json
```

```python
df = pd.read_csv("../Datasets/travel.csv")

print("Shape:", df.shape)
df.head()
```

```python
df.info()
df.describe(include='all')
```

```python
useful_cols = [
    "Zone",
    "State",
    "City",
    "Name",
    "Type",
    "Google review rating",
    "Best Time to visit",
    "time needed to visit in hrs",
    "Airport with 50km Radius"
]

df = df[useful_cols]
```

```python
df.info()
df.head()
```

```python
df['City'] = df['City'].replace({
    'New Delhi': 'Delhi'
})
df.isnull().sum()
```

```python
print(df['City'].unique())
```

```python
df['City'] = df['City'].str.strip().str.title()
df['State'] = df['State'].str.strip().str.title()
df['Type'] = df['Type'].str.strip().str.title()
df['Zone'] = df['Zone'].str.strip().str.title()
```

```python
unique_types = df['Type'].unique()
print("Unique Types:", unique_types)
print("Total unique types:", len(unique_types))
```

```python
type_mapping = {

    # Nature
    'Park': 'Nature',
    'National Park': 'Nature',
    'Wildlife Sanctuary': 'Nature',
    'Bird Sanctuary': 'Nature',
    'Botanical Garden': 'Nature',
    'Lake': 'Nature',
    'Waterfall': 'Nature',
    'Valley': 'Nature',
    'Hill': 'Nature',
    'Mountain Peak': 'Nature',
    'Scenic Area': 'Nature',
    'Scenic Point': 'Nature',
    'Viewpoint': 'Nature',
    'Island': 'Nature',
    'River Island': 'Nature',
    'Natural Feature': 'Nature',
    'Cave': 'Nature',
    'Dam': 'Nature',
    'Orchard': 'Nature',
    'Vineyard': 'Nature',
    'Tea Plantation': 'Nature',
    'Sunrise Point': 'Nature',
    'Gravity Hill': 'Nature',
    'Confluence': 'Nature',
    'Village': 'Nature',
     'Beach':'Nature',

    # Historical
    'Fort': 'Historical',
    'Palace': 'Historical',
    'Tomb': 'Historical',
    'Tombs': 'Historical',
    'Mausoleum': 'Historical',
    'War Memorial': 'Historical',
    'Memorial': 'Historical',
    'Monument': 'Historical',
    'Stepwell': 'Historical',
    'Prehistoric Site': 'Historical',
    'Rock Carvings': 'Historical',
    'Historical': 'Historical',
    'Border Crossing': 'Historical',
     'Museum': 'Historical',
    'Science': 'Historical',
    'Observatory': 'Historical',
    'Landmark': 'Historical',
    'Sculpture Garden': 'Historical',
    'Government Building': 'Historical',
    'Site': 'Historical',
    'Cultural':'Historical',

    # Religious
    'Temple': 'Religious',
    'Temples': 'Religious',
    'Church': 'Religious',
    'Mosque': 'Religious',
    'Gurudwara': 'Religious',
    'Monastery': 'Religious',
    'Shrine': 'Religious',
    'Religious Shrine': 'Religious',
    'Religious Site': 'Religious',
    'Religious Complex': 'Religious',
    'Spiritual Center': 'Religious',
    'Ghat': 'Religious',

    # Adventure
    'Adventure Sport': 'Adventure',
    'Trekking': 'Adventure',
    'Ski Resort': 'Adventure',
    'Suspension Bridge': 'Adventure',

    # Urban
    'Market': 'Urban',
    'Mall': 'Urban',
    'Commercial Complex': 'Urban',
    'Entertainment': 'Urban',
    'Theme Park': 'Urban',
    'Amusement Park': 'Urban',
    'Aquarium': 'Urban',
    'Zoo': 'Urban',
    'Cricket Ground': 'Urban',
    'Race Track': 'Urban',
    'Urban Development Project': 'Urban',
    'Film Studio': 'Urban',
    'Township': 'Urban',
    'Promenade': 'Urban',
    'Bridge': 'Urban'
}
```

```python
df['Category'] = df['Type'].map(type_mapping)
print("Unmapped rows:", df['Category'].isnull().sum())
```

```python
category_index = {
    "Nature": 0,
    "Historical": 1,
    "Religious": 2,
    "Adventure": 3,
    "Urban": 4
}
df['Category_ID'] = df['Category'].map(category_index)
df[['Type', 'Category', 'Category_ID']].head(10)
```

```python
zone_vectors = pd.crosstab(df['Zone'], df['Category'])
zone_vectors = zone_vectors.div(zone_vectors.sum(axis=1), axis=0)
zone_vectors.head()
```

```python
state_vectors = pd.crosstab(df['State'], df['Category'])
state_vectors = state_vectors.div(state_vectors.sum(axis=1), axis=0)
state_vectors.head()
```

```python
cols = ['Nature', 'Historical', 'Religious', 'Adventure', 'Urban']

for col in cols:
    if col not in zone_vectors:
        zone_vectors[col] = 0
    if col not in state_vectors:
        state_vectors[col] = 0
```

```python
zone_dict = zone_vectors.to_dict(orient='index')
state_dict = state_vectors.to_dict(orient='index')
```

```python
with open("../Datasets/zone_vectors.json", "w") as f:
    json.dump(zone_dict, f, indent=4)

with open("../Datasets/state_vectors.json", "w") as f:
    json.dump(state_dict, f, indent=4)
```

```python
city_counts = pd.crosstab(df['City'], df['Category'])
for col in ['Nature', 'Historical', 'Religious', 'Adventure', 'Urban']:
    if col not in city_counts:
        city_counts[col] = 0
city_counts = city_counts.reset_index()
city_counts.head()
```

```python
cols = ['Nature', 'Historical', 'Religious', 'Adventure', 'Urban']

city_vectors = city_counts.copy()

city_vectors[cols] = city_vectors[cols].div(
    city_vectors[cols].sum(axis=1), axis=0
)
```

```python
city_vectors.head()
```

```python
city_rating = df.groupby('City')['Google review rating'].mean().reset_index()
```

```python
city_vectors = city_vectors.merge(city_rating, on='City')
city_vectors['rating_norm'] = (
    (city_vectors['Google review rating'] - 3) / 2
)
```

```python
city_dict = city_vectors.set_index('City').to_dict(orient='index')

with open("../Datasets/city_vectors.json", "w") as f:
    json.dump(city_dict, f, indent=4)
```
